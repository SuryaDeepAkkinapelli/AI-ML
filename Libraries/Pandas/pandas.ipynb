# Dataframes in Pandas

```python
# A dataframe can be created in 3 ways

# By using dictionary of lists
info = {
    "Name" : ["Rahul","Virat","Rohit"],
    "Age" : [34, 37, 39],
    "Rating":[910, 923, 932]
}
df = pd.DataFrame(info)
print(df,type(df))
print(df.index) #RangeIndex(start=0, stop=3, step=1)
print(df.columns) #Index(['Name', 'Age', 'Rating'], dtype='object')


# By using list of lists
df = pd.DataFrame([["Surya",18],["Srihari",20],["Rahul",34]], columns = ["Name", "Age"])
print(df)

# By using numpy Arrays
arr = np.array([[1,2,3],[4,5,6]])
df = pd.DataFrame(arr, columns = ["A","B",'C'])
print(df)
```

# Melt & Pivot

```python
df = pd.DataFrame({
    'country': ['USA','USA','India','India'],
    'year':[2020,2021,2020,2021],
    'sales' :[100,120,90,110],
    'profit':[20,25,18, 22]
})

# wide to long
melted = df.melt(
    id_vars = ['country','year'], # keep
    value_vars = ['sales','profit'], # what to melt
    var_name = 'metric', # name of new col
    value_name = 'value' # name of the val column
)

#long to wide
melted.pivot(
    index = ['country','year'],
    columns = ['metric'],
    values = ['value']
)
```

# Merge & Join data

```python
df_cust = pd.DataFrame({
    'customer':['srihari','Ajay','Nitish','Rahul'],
    'customer_id': [1,2,3,4],
    
}) 
df_orders = pd.DataFrame({
    "order_id":[101,102,103,104],
    "customer_id":[2,1,4,5],
    "amount":[100,200,102,150]
})
# pd.merge(df_cust,df_orders,on='customer_id')
pd.merge(df_cust,df_orders,on='customer_id',how = 'left')
pd.merge(df_cust,df_orders,on='customer_id',how = 'right')
pd.merge(df_cust,df_orders,on='customer_id',how = 'outer')
```

# Data concatenation

```python
df1 = pd.DataFrame({
    'id' : [1,2,3],
    'name':['a','b','c']
})
df2 = pd.DataFrame({
    'id' : [4,5,6],
    'name':['x','y','z']
})

pd.concat([df1,df2]) # row concat(two tables stack each other) (label preserved)
pd.concat([df1,df2],ignore_index = True) # row concat (label not preserved)

pd.concat([df1,df2],axis = 1) # col(joined side by side) concat
```
