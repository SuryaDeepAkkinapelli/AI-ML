```python
# Data collection
import json
def load_data(filename):
    with open(filename,"r") as f:
        data = json.load(f)
    return data
data = load_data("store-data.json")
# print(data)
```

```python
#Data Cleaning
def clean_data(d):
    text_to_nums = {
        "one" : 1,
        "two" : 2,
        "three" : 3,
        "four" : 4,
        "five" : 5
                   }
    cleaned_data = []
    unique_users = set()
    for user in d:
        rating = str(user["rating"]).strip().lower()
        if(rating in text_to_nums): # four -> 4
            user["rating"] = text_to_nums[rating]
            print(user["rating"])

        #Handling missing values
        if(user.get("age") == None):
            user["age"] = None

        #De duplication
        if(user["name"].strip() not in unique_users):
            cleaned_data.append(user)
            unique_users.add(user["name"].strip())
    return cleaned_data
```

```python
cd = clean_data(data)
for i in cd:
    print(i)
```

```python
def create_insights(data):
    avg_rating = 0
    poor_rating = 0
    for user in data:
        avg_rating += float(user["rating"])
        if(float(user["rating"]) < 3):
            poor_rating+=1
    print(f"Average rating = {avg_rating / len(data)}")
    print(f"Percentage of poor ratings = {(poor_rating / len(data)) * 100}%")
        
```

```python
create_insights(cd)
```

```python
def recommendation_feature(data):
    recommendations = []
    for user in data:
        curr_recom = {}
        curr_recom["name"] = user["name"]
        if(float(user["rating"]) >= 4):
            curr_recom["brand"] = "Apple"
        else:
            curr_recom["brand"] = "Pixel"
        recommendations.append(curr_recom)
    return recommendations
```
